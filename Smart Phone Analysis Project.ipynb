# Exploratory Data Analysis (EDA)

# Univariate Analysis (Single Column)

```python
# Distribution of prices
# RAM sizes frequency
# OS versions count

df['price'].hist()
df['ram'].value_counts().plot(kind='bar')
```

# B. Bivariate Analysis (Compare 2 Variables)

```python
# Price vs Spec Score
# RAM vs Performance
# Battery vs Charging Speed

import matplotlib.pyplot as plt

plt.scatter(df['price'], df['spec_score'])
plt.xlabel("Price")
plt.ylabel("Spec Score")
plt.show()
```

# 3. Grouping & Aggregations

# #A. By Brand

```python
df.groupby('brand_name')['price'].mean().sort_values()
```

## B. By Price Category

```python
df.groupby('price_category')[['ram', 'battery_capacity(mAh)']].mean()
```
